```python
from pathlib import Path
from IPython.display import HTML, display

image_folder = Path("./data/images")  # Example: Path("./images")

image_files = list(image_folder.glob("*.jpg")) + list(image_folder.glob("*.jpeg")) + list(image_folder.glob("*.png"))

html = """
<style>
.img-container {
  display: inline-block;
  margin: 5px;
  position: relative;
}

.img-container img {
  width: 100px;
}
</style>
<div>
"""

for img_path in image_files:
    html += f"<div class='img-container'><img src='{img_path.as_posix()}'></div>"

html += "</div>"

display(HTML(html))
```
